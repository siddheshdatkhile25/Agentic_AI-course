```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq

load_dotenv()
```

```python
llm = ChatGroq(
    model="openai/gpt-oss-120b",
)

response = llm.invoke("What are major topics of system Design")
print(response.text)
```

```python
response = llm.invoke([
    ["system" , "You are Helpful guide that answer in short and simple way in two lines"],
    ["human" , "Important topics to cover in system Design as 1 year of experience candidate"]
])
print(response.text)
```

```python
prompt = "write a poem for a life of a bachelor in pune life in hindi but wording should be english"
llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)
response = llm.invoke(prompt)
print(response.text)
```
